# Adding The Images

```python
compressed_image = cv2.resize(image,(300,300))
cv2.imshow("COMPRESSED IMAGE",compressed_image)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
gray_image = cv2.cvtColor(compressed_image,cv2.COLOR_RGB2GRAY)
cv2.imshow("GRAYSCALE IMAGE",gray_image)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
hls_image = cv2.cvtColor(compressed_image,cv2.COLOR_RGB2HLS)
cv2.imshow("HLS IMAGE",hls_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
hsv_image = cv2.cvtColor(compressed_image,cv2.COLOR_RGB2HSV)
cv2.imshow("HSV IMAGE",hsv_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Creating The Image

```python
black = np.zeros((800,800),dtype='int8')
black
```

```python
cv2.imshow("BLACK",black)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
black = np.zeros((600,800,3),dtype='uint8')
cv2.rectangle(img = black,
              pt1 = (100,100),
              pt2 = (400,300),
              color = (255,0,0),
              thickness = -1 )
cv2.imshow("BLACK",black)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
black = np.zeros((600,800,3),dtype='uint8')
cv2.circle(img = black,
           center = (250,300),
          radius = 100,
          color = (0,0,255),
          thickness= 300)
cv2.imshow("BLACK",black)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
