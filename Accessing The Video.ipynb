# Reading All The Frames

```python
Video = cv2.VideoCapture("C:/Users/Kadikondi Gajendra/Downloads/video.mp4")

while True:
    success,frame = Video.read()
    if not success:
        break
        
    cv2.imshow("Frames",frame)
    if cv2.waitKey(100) == ord('a'):
        Video.release()
        cv2.destroyAllWindows()
```

```python
Video = cv2.VideoCapture(r"C:/Users/Kadikondi Gajendra/Downloads/video.mp4")

while True:
    success,frame = Video.read()
    if not success:
        break
    cv2.imshow("Frames",cv2.cvtColor(frame,cv2.COLOR_RGB2HLS))
    if cv2.waitKey(100) == ord('a'):
        Video.release()
        cv2.destroyAllWindows()
```

# Acessing Web Cam

```python
Video = cv2.VideoCapture(0)

while True:
    success,frame = Video.read()
    if not success:
        break

    cv2.putText(img = frame,
                text = "Face Detected",
                org = (100,100),
                fontFace = cv2.FONT_HERSHEY_TRIPLEX,
                fontScale = 2,
                color = (100,24,34),
                thickness = 2)

    cv2.rectangle(img = frame,
             pt1 = (250,150),
             pt2 = (500,400),
             color = (0,255,0),
              thickness = 2)
        
    cv2.imshow("Frames",frame)
    if cv2.waitKey(100) == ord('a'):
        Video.release()
        cv2.destroyAllWindows()
```
